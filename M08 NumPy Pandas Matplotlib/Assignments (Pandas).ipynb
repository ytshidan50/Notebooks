###### 7. Count the number of missing values in each column.

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', None, 'David'],
    'Age': [25, None, 35, 40],
    'City': ['NY', 'LA', None, None]
})
print(df)
missing_val=df.isnull().sum() # To count missing values in each column
print(missing_val)

full_missing=df.isnull().sum().sum() # To count total missing values in the dataframe
print(full_missing)
```

###### 8. Replace all missing values with the column mean.

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', None, 'David'],
    'Age': [25, None, 35, 40],
    'Salary': [50000, 60000, None, 70000]
})
print(df,"\n")
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
print(df)
```

```python
# One Line Solution
df= pd.DataFrame({
    'Name' : ['Alice', 'Bob', None, 'David'],
    'Age' : [25, None, 35, 40],
    'Salary' : [50000, 60000, None, 70000]
})
print(df)

df=df.fillna(df.mean(numeric_only=True)) # One Line Solution
print(df)
```

###### 13. Merge two DataFrames based on a common column.

```python
df1 = pd.DataFrame({
    'EmployeeID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

df2 = pd.DataFrame({
    'EmployeeID': [1, 2, 3, 5],
    'Salary': [50000, 60000, 70000, 80000]
})

merged=pd.merge(df1, df2, on='EmployeeID')
print(merged)
```

```python
left_merge=pd.merge(df1, df2, on='EmployeeID', how='left')
print(left_merge)
```

```python
right_merge=pd.merge(df1, df2, on='EmployeeID', how='right')
print(right_merge)
```

```python
outer_merge=pd.merge(df1, df2, on='EmployeeID', how='outer')
print(outer_merge)
```

###### 14. Concatenate two DataFrames vertically and horizontally.

```python
# Vertical Concatenation
df1 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

df2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

vertical=pd.concat([df1, df2], axis=0)
print(vertical)
```

```python
# Horizontal Concatenation
df1 = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

df2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})
horizontal=pd.concat([df1, df2], axis=1)
print(horizontal)
```

###### 15. Use apply() to create a new column based on custom logic.

```python
def assign_grade(mark):
    if mark > 90:
        return 'A'
    elif mark > 80:
        return 'B'
    elif mark > 70:
        return 'C'
    elif mark > 60:
        return 'D'
    else:
        return 'E'
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Score': [85, 42, 73, 91]
})

df['Grade']=df['Score'].apply(assign_grade)
print(df)
```

###### 16. Reshape data using pivot() and melt().

```python
# Pivot
df = pd.DataFrame({
    'Name': ['Alice', 'Alice', 'Bob', 'Bob'],
    'Month': ['Jan', 'Feb', 'Jan', 'Feb'],
    'Sales': [200, 250, 300, 280]
})
print(df, "\n")
pivoted=df.pivot(index='Name', columns='Month', values='Sales') # Spread rows into columns
print(pivoted)
```

```python
# Melt
df = pd.DataFrame({
    'Student': ['A', 'B', 'C'],
    'Math': [85, 90, 78],
    'Science': [92, 88, 80],
    'English': [75, 82, 89]
})
print(df,'\n')
melted=df.melt(id_vars='Student', var_name='Subject', value_name='Score') # Melt makes data longer
print(melted)
```

###### 18. Replace all outliers in a column with the column's median.

```python
df = pd.DataFrame({
    'Score': [10, 12, 14, 200, 15, 13, 250, 11, 14]
})
print(df)
q1=df['Score'].quantile(0.25)
q3=df['Score'].quantile(0.75)
iqr=q3-q1
lower_bound=q1 - 1.5 * iqr
upper_bound=q3 + 1.5 * iqr
median=df['Score'].median()
df['Score'] = df['Score'].apply(lambda x: median if (x < lower_bound or x > upper_bound) else x)
print(df)
```

######  19. Create a new column that categorizes numerical values into bins using pd.cut()

```python
df = pd.DataFrame({
    'Age': [12,90 , 18, 25, 32, 45, 52, 67, 80]
})
print(df)
bins=[0 , 17, 30, 50, 100]
labels=['Child', 'Young Adult', 'Adult', 'Senior']
df['Category']=pd.cut(df['Age'], bins=bins, labels=labels)
print(df)
```

```python
df = pd.DataFrame({
    'Age': [10, 20, 30, 40, 50, 60]
})
print(df)
df['Age_Group']=pd.cut(df['Age'], bins=3) # Equal Width Bins
print(df)
```
