```python
# 1d array 10 random, below mean 0, above mean 1
np.random.seed(0)
arr=np.random.choice(10, size=10, replace=False)

print("Original Array:", arr)

# Compute mean
mean_val = arr.mean()
print("Mean:", mean_val)

# Convert values based on condition
# result = np.where(arr > mean_val, 0, 1)
arr[arr < mean_val] = 0
arr[arr > mean_val] = 1

print("Modified Array:", arr)

```

```python
# Boston Celtics salary > average salary of their Team, but < average salary of the whole dataset
import pandas as pd

df = pd.read_csv("../Datasets/nba.csv")
df
```

```python
glob_avg=df['Salary'].mean()
avg_sal=df.groupby('Team')['Salary'].mean()
print(avg_sal)
avg_sal_boston = avg_sal.loc['Boston Celtics']
hi=df[df['Salary'].map(lambda x:x > avg_sal_boston and x < glob_avg)]
print(hi)
wizards = hi[hi['Team'] == 'Washington Wizards']
print(wizards)
```

```python
import matplotlib.pyplot as plt

salary_dist=df.groupby('Team')['Salary'].sum().reset_index()
print(salary_dist)
# plt.figure()
plt.bar(salary_dist['Team'], salary_dist['Salary'])

```

```python
import pandas as pd
import numpy as np

# Create a pandas Series of 10 random numbers
s = pd.Series(np.random.rand(10))
print("Original Series:")
print(s)

# Calculate mean
mean_val = s.mean()
print("\nMean:", mean_val)

# Replace values based on condition
s_modified = s.apply(lambda x:0 if x > mean_val else 1)

print("\nModified Series:")
print(s_modified)
```

```python
x=np.array(['Monday','Tuesday','Wednessday','Thursday','Friday','Saturday'])
y1=np.random.randint(10,60,size=6)
y2=np.random.randint(10,60,size=6)
y3=np.random.randint(10,60,size=6)
properties=dict(
    linestyle='solid',
    linewidth=2,
    marker='.',
    markerfacecolor='#5f5f5f',
    markeredgecolor='#000000',
    markersize=8
)
plt.title("Daily Sale rate - 08th Week.",color='red',family='Arial',fontweight='bold',fontsize='18')
plt.xlabel("Days",color='black',family='Arial',fontsize='14')
plt.ylabel("Count",color='black',family='Arial',fontsize='14')
plt.tick_params(axis='both',color='gray',colors='gray')
plt.plot(x,y1,**properties,color='#ff0000')
plt.plot(x,y2,**properties,color='#00ff00')
plt.plot(x,y3,**properties,color='#0000ff')
plt.grid(axis='both',color='red',linewidth=0.4,linestyle='dotted')
plt.show()
```
