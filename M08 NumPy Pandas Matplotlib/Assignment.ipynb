```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
print('Started')
```

```python
years=np.array([2021,2022,2023,2024,2025])
income=np.array([100000,120000,112000,100000,120000])
outgo=np.array([40000,35000,42000,37500,41000])
incharge=np.array(['John','Andro','John','Jack','Andro'])
df=pd.DataFrame({"Incharge":incharge,'Income':income,'Outgo':outgo},index=years)
df
```

```python
df['Profit']=df['Income']-df['Outgo']
df
```

```python
line_style=dict(
    linestyle='solid',
    linewidth=2,
    marker='.',
    markerfacecolor='gray',
    markeredgecolor='black',
    markersize=10
)
plt.grid(axis='both',color='gray',linewidth=0.3,linestyle='dashed')
plt.tick_params(axis='x',colors='brown')
plt.tick_params(axis='y',colors='black')
plt.tick_params(color='gray',axis='both')
plt.plot(years,income,**line_style,color='green')
plt.plot(years,outgo,**line_style,color='orange')
plt.plot(years,df['Profit'],**line_style,color='lime')
plt.xticks(years)
plt.title("Business Developments",color='red',fontweight='bold',fontsize=18)
plt.xlabel("Years",color='blue',fontsize=14)
plt.ylabel('Amount',color='black',fontsize=14)
plt.show()
```

```python
incharge_reports = df.groupby("Incharge").size().reset_index(name='Count')
plt.bar(incharge_reports['Incharge'],incharge_reports['Count'],color='skyblue')
plt.title('Incharge Comparison',fontsize=18,color='blue',fontweight='bold')
plt.xlabel('Persons',color='black')
plt.ylabel('Count of being Incharge')
plt.yticks(incharge_reports['Count'])
plt.tick_params(color='gray')
```

```python
# pie chart is the visual representation of slices as each with total of 100%
plt.pie(
        incharge_reports["Count"],
        labels=incharge_reports["Incharge"],
        autopct="%1.1f%%",
        colors=["red", "blue", "green"],
        explode=[0, 0, 0.2],
        shadow=True,
        startangle=90,
)
```

```python
studentsstats=pd.DataFrame({'Study Hours':[0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8],
                            'Marks':[55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]})
studentsstats
```

```python
# scatter plots visually represents the correlation between the data
plt.scatter(studentsstats['Study Hours'],
            studentsstats['Marks'],
            color='skyblue',
            alpha=0.5,
            label='Class A'
)
x2 = np.array([0, 1, 2, 2,3, 4,5,6,7,8,8])
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97])
plt.scatter(x2,y2,
            color='red',
            alpha=0.5,
            label='Class B'
)
plt.ylabel("Marks")
plt.xlabel('Hours Studied')
plt.title("Stats of students with the study time")
plt.legend()
plt.show()
```

```python
# Histograms is the visual respresentation of the distribution of quantitive data
score=np.random.normal(loc=80,scale=10,size=100)
score=np.clip(score,0,100)
plt.hist(score,bins=20,color='lightgreen',edgecolor='black')
```
