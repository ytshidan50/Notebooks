# Eigenvalues & Eigenvectos

## Finding the eigenvalues

```python
mxv1=[[4, 1], [2, 3]]
(a,b),(c,d)=mxv1
trace=a+d
det=a*d-b*c
lam1=(trace+sqrt(trace**2-4*det))/2
lam2=(trace-sqrt(trace**2-4*det))/2
print('Eigenvalues:',lam1,lam2)
```

## Creating the eigenvectors

```python
evec=[]
for lam in (lam1,lam2):
    x=b
    y=lam-a
    evec.append([x,y])
for i in range(2):
    print(f'eigenvector {evec[i]} for eigenvalue {[lam1,lam2][i]}')
```

## Diagonalization Matrix

```python
eigenvalues,eigenvectors=np.linalg.eig(mxv1)
print(eigenvalues)
print(eigenvectors)
```

### Creating the vector matrix P

```python
P=eigenvectors
P
```

### Creating the diagonal matrix of eigenvalues D

```python
t=[]
for i in range(len(eigenvalues)):
    t.append(0)
t=[t.copy(),t.copy()]
for i in range(len(eigenvalues)):
    t[i][i]=eigenvalues[i]
D=t.copy()
D
```

### Creting the inverse of P P_inv

```python
P_inv=np.linalg.inv(P)
P_inv
```

```python
PD=np.matmul(P,D)
PDP_inv=np.matmul(PD,P_inv)
PDP_inv
```

## Covariance and PCA

```python
data = [
    [170, 65],
    [172, 67],
    [168, 64],
    [180, 75]
]
```

```python
h_sum=0
w_sum=0
for (h,w) in data:
    h_sum+=h
    w_sum+=w
h_mean=h_sum/len(data)
w_mean=w_sum/len(data)
data_cen=[]
for (h,w) in data:
    h_cen=h-h_mean
    w_cen=w-w_mean
    data_cen.append([h_cen,w_cen])
data_cen
```

```python
co_xx=0
co_xy=0
co_yy=0
for (h_cen,w_cen) in data_cen:
    co_xx+=h_cen*h_cen
    co_xy+=h_cen*w_cen
    co_yy+=w_cen*w_cen
cov_xx=co_xx/(len(data_cen)-1)
cov_xy=co_xy/(len(data_cen)-1)
cov_yy=co_yy/(len(data_cen)-1)
cov=[
    [cov_xx, cov_xy],
    [cov_xy, cov_yy]
]
cov
```

### PCA

```python
trace=cov_xx+cov_yy
det=cov_xx*cov_yy-cov_xy*cov_xy
lam1=(trace+sqrt(trace**2-4*det))/2
lam2=(trace-sqrt(trace**2-4*det))/2
eigenvalues=[lam1,lam2]
eigenvectors=[]
for lam in eigenvalues:
    eigenvectors.append([cov_xy,lam-cov_xx])
if lam1>lam2:
    PC1=eigenvectors[0]
else:
    PC1=eigenvectors[1]
projected=[]
for (h_cen,w_cen) in data_cen:
    projected.append(h_cen*PC1[0]+w_cen*PC1[1])
projected
```

## Gradient Descent Optimization

```python
def f(x):
    return x**2
def grad(x):
    return 2*x

lr=0.1
x=5

x_range=[i/10 for i in range(-50,50)]
y_range=[f(i) for i in x_range]
plt.plot(x_range,y_range,label="f(x) = x^2")

x_value=[]
y_value=[]
for i in range(20):
    x_value.append(x)
    y_value.append(f(x))
    g=grad(x)
    x=x-lr*g
    print(f'{x} {f(x)}')

plt.scatter(x_value,y_value,label="Gradient descent steps")
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()
plt.title("Gradient descent optimization")

plt.show()
```
