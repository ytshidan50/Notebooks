```python
import pandas as pd
df=pd.read_csv('iris.csv')
df.head()

df=df.drop('Id',axis=1)
```

```python
df.head()
```

```python
import seaborn as sns
sns.pairplot(data=df,hue='Species')
```

```python
sns.pairplot(data=df,vars=['PetalLengthCm','PetalWidthCm'],hue='Species',diag_kind='hist',markers=['o','s','^'])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import OneHotEncoder
oh=OneHotEncoder(sparse_output=False)
encoded_df=oh.fit_transform(df[['Species']])
encoded_df=pd.DataFrame(encoded_df,columns=oh.get_feature_names_out(['Species']))
encoded_df
```

```python
corr=encoded_df.corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
df=pd.read_csv('iris.csv')
df.head()
```

```python
df['Species'].unique()
```

```python
df['Species']=df['Species'].map({"Iris-setosa":0,
                                 'Iris-versicolor':1,
                                 "Iris-virginica":2})
```

```python
df.head()
```

```python
df=df.drop("Id",axis=1)
```

```python
cov_mat=df.cov()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(data=cov_mat,annot=True,cmap='coolwarm')
plt.show()
```
