```python
import pandas as pd
df=pd.read_csv('movie.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.drop(columns='homepage',inplace=True)
```

```python
df=df.dropna(subset=['overview'])
```

```python
df['release_date']=df['release_date'].fillna(df['release_date'].mode()[0])
```

```python
df=df.drop('tagline',axis=1)
```

```python
df.describe()
```

```python
num_df=df.select_dtypes(include='number')
cat_df=df.select_dtypes(include='object')
```

```python
num_df.columns
```

```python
cat_df.columns
```

```python
num_df=num_df.drop('id',axis=1)
```

```python
num_df.columns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))

plt.subplot(2,3,1)
sns.kdeplot(num_df['budget'])

plt.subplot(2,3,2)
sns.kdeplot(num_df['popularity'])

plt.subplot(2,3,3)
sns.kdeplot(num_df['revenue'])

plt.subplot(2,3,4)
sns.kdeplot(num_df['runtime'])

plt.subplot(2,3,5)
sns.kdeplot(num_df['vote_average'])

plt.subplot(2,3,6)
sns.kdeplot(num_df['vote_count'])


plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15,8))

for i,col in enumerate(num_df.columns,1):
    plt.subplot(2,3,i)
    sns.kdeplot(num_df[col])
    plt.title([col])
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15,8))

for i,col in enumerate(num_df.columns,1):
    plt.subplot(2,3,i)
    sns.histplot(num_df[col],kde=True,color='Green',edgecolor='black')
    plt.title([col])
plt.tight_layout()
plt.show()
```

```python
cat_df.nunique()
```

```python
num_df.nunique()
```

```python
for i,col in enumerate(num_df.columns,1):
    Skewness=num_df[col].skew().round(2)
    print(f'{col}:{Skewness}')
```

```python
for i,col in enumerate(num_df.columns,1):
    Kurtosis=num_df[col].kurt().round(2)
    print(f'{col}:{Kurtosis}')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))

for i,col in enumerate(num_df.columns,1):
    plt.subplot(2,3,i)
    sns.boxplot(num_df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```
