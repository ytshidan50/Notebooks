```python
plt.figure(figsize=(10,8))
plt.grid(True)
plt.scatter(df['PetalLengthCm'],df['PetalWidthCm'],color='Green',alpha=0.6,edgecolors='black')
plt.title("Relationship between petal length and petal width")
plt.xlabel("Petal Length(Cm)")
plt.ylabel("Petal Width (Cm)")
```

```python
import matplotlib.pyplot as plt
plt.grid(True)
plt.bar(df['Species'],50,color='Skyblue',edgecolor='Black',width=0.5,alpha=0.7)
plt.xlabel("Flower Type")
plt.ylabel("Number Of Flowers")
plt.show()
```

```python
mean_df=df.groupby("Species")['SepalLengthCm'].agg('mean')
mean_df
```

```python
plt.bar(mean_df.index,mean_df)
plt.xlabel("Species")
plt.ylabel("Sepal Length")
for i, value in enumerate(mean_df):
    plt.text(i, value + 0.05, f"{value:.2f}", ha="center")
```

```python
mean_sw=df.groupby("Species")['SepalWidthCm'].agg('mean')
mean_sw
```

```python
plt.bar(mean_sw.index,mean_sw)
plt.xlabel("Species")
plt.ylabel("Sepal width")
for i,value in enumerate(mean_sw):
    plt.text(i,value+0.05,f'{value:.2f}',ha='center')
```

```python
import pandas as pd
df=pd.read_csv("iris.csv")

df.head()
```

```python
df['Species'].unique()
```

```python
Setosa=df[df['Species']=='Iris-setosa']['PetalWidthCm']
Versicolor=df[df['Species']=='Iris-versicolor']['PetalWidthCm']
Virginica=df[df["Species"]=='Iris-virginica']['PetalWidthCm']
```

```python
plt.hist(Setosa,label='Setosa')
plt.hist(Versicolor,label='Versicolor')
plt.hist(Virginica,label='Virginica')
plt.title("Petal Width of 3 types")
plt.xlabel("Petal Width")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
