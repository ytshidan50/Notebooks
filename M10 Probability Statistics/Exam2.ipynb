```python
import numpy as np
from scipy import stats
```

```python
# data=np.random.arange([1,0],size=100)
data=[34,35,24,27,54,34,47,39]
print(data)
```

```python
print('Probability of Head',sum(1 for x in data if x==1)/len(data))
print('Probability of Tail',sum(1 for x in data if x==0)/len(data))
```

```python
mean=np.mean(data)
mu=40
n=len(data)
s=np.std(data,ddof=1)
t_stat=(mean-mu)/(s/np.sqrt(n))
p_value=2*(1-stats.t.cdf(data,df=n-1))
print('t_stat',t_stat)
print('p_value',p_value)
```
