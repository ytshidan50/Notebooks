```python
import numpy as np
from collections import Counter
print('started')
```

```python
data = [12, 15, 18, 20, 22, 25]
mean=sum(data)/len(data)
median=(data[len(data)//2]+data[len(data)//2-1] )/2  if len(data)%2==0 else data[len(data)//2]
freq=Counter(data)
biggest=1
for count in freq.values():
    if count>biggest:
        biggest=count
mode=[]
for value,count in freq.items() if biggest>1 else []:
    if count == biggest:
        mode.append(value)
print(mean)
print(median)
print(mode)
```

```python
big=data[0]
sm=999
for i in range(len(data)):
    if data[i]>big:
        big=data[i]
    elif data[i]<sm:
        sm=data[i]
print(f'range {big-sm}')
```
