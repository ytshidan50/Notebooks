# Queue

```python
class Queue:
    def __init__(self, capacity):
        self.q = []
        self.capacity = capacity

    def isEmpty(self):
        return len(self.q) == 0

    def isFull(self):
        return self.capacity == len(self.q)

    def front(self):
        return f"[Front] Queue underflow!" if self.isEmpty() else f"[Front] {self.q[0]}."
    
    def rear(self):
        return f"[Rear] Queue underflow!" if self.isEmpty() else f"[Rear] {self.q[-1]}."

    def enqueue(self, val):
        if self.isFull():
            return f"[Enqueue] Queue overflow! could not insert {val}!"
        else:
            self.q.append(val)
            return f"[Enqueue] {val} inserted to the queue."

    def dequeue(self):
        if self.isEmpty():
            return f"[Dequeue] Queue underflow!"
        else:
            return f"[Dequeue] {self.q.pop(0)} has been successfully removed from the queue."

    def size(self):
        return f"[Size] {len(self.q)}."

    def display(self):
        if self.isEmpty():
            return f"[Display] Queue underflow!"
        else:
            return f"[Display] Queue: {self.q}."
```
