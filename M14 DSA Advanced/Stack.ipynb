# Stack

```python
class Stack:
    def __init__(self, capacity):
        self.st = []
        self.capacity = capacity

    def isEmpty(self):
        return len(self.st) == 0

    def isFull(self):
        return len(self.st) == self.capacity

    def push(self, val):
        if self.isFull():
            return f"[Push] Stack overflow!"
        else:
            self.st.append(val)
            return f"[Push] {val} pushed to stack."

    def pop(self):
        if self.isEmpty():
            return f"[Pop] Stack underflow!"
        else:
            return f"[Pop] {self.st.pop()} popped from the stack."

    def size(self):
        return len(self.st)

    def peek(self):
        if self.isEmpty():
            return f"[Peek] Stack underflow!"
        else:
            return f"[Peek] {self.st[-1]}."

    def bottom(self):
        if self.isEmpty():
            return f"[Bottom] Stack underflow!"
        else:
            return f"[Bottom] {self.st[0]}."

    def display(self):
        return f"[Display] Stack: {self.st}." if not self.isEmpty() else f"[Display] Stack underflow!"
```

```python
obj = Stack(6)
print(obj.display())
print(obj.push(48))
print(obj.push(54))
print(obj.push(78))
print(obj.push(25))
print(obj.push(75))
print(obj.push(10))
print(obj.push(23))
print("Size:", obj.size())
print("Full:", obj.isFull())
print(obj.display())
print(obj.pop())
print(obj.peek())
print(obj.pop())
print(obj.push(23))
print("Size:", obj.size())
print("Full:", obj.isFull())
print(obj.display())
print(obj.pop())
print(obj.pop())
print(obj.pop())
print(obj.pop())
print(obj.pop())
print(obj.pop())
print("Empty:", obj.isEmpty())
```
