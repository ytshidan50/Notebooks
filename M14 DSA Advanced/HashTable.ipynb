# Hash Table

## Chaining Method

```python
class HashTableChaining:
    def __init__(self):
        self.size = 10
        self.table = [[] for _ in range(self.size)]

    def HashFunction(self, key):
        return key%self.size

    def insert(self, key):
        i = self.HashFunction(key)
        if key not in self.table[i]:
            self.table[i].append(key)
            print(f"[Insertion] {key} inserted at {i}th index")
        else:
            print(f"[Insertion] {key} already exist in {i}")

    def search(self, key):
        i = self.HashFunction(key)
        if key in self.table[i]:
            print(f"[Search] Found {key} in {i}th index")
        else:
            print(f"[Search] {key} not found")

    def delete(self, key):
        i = self.HashFunction(key)
        if key in self.table[i]:
            self.table[i].remove(key)
            print(f"[Deletion] Successfully deleted {key} from {i}th index")
        else:
            print(f"[Deletion] {key} not found to delete")

    def display(self):
        print("\n Hash Table")
        for i in range(self.size):
            print(f"{i}: {self.table[i]}")
```

```python
obj = HashTableChaining()
obj.search(10)
obj.insert(10)
obj.search(10)
obj.insert(34)
obj.insert(85)
obj.insert(45)
obj.insert(32)
obj.insert(9)
obj.insert(90)
obj.insert(32)
obj.insert(27)
obj.insert(86)
obj.search(45)
obj.delete(78)
obj.delete(45)
obj.search(45)
obj.display()
```

## Open Addressing

```python
class HashTableOpenAddressing:
    def __init__(self):
        self.size = 10
        self.table = [None]*self.size
    def HashFunction(self, key):
        return key%self.size

    def insert(self, key):
        i = self.HashFunction(key)
        original_index = i
        while self.table[i] is not None:
            if self.table[i] == key:
                print(f"[Insertion] Duplicate is not allowed!")
                return
            i = (i+1)%self.size
            if i == original_index:
                print(f"[Insertion] Index overflow.")
                return
        self.table[i] = key
        print(f"[Insertion] Successfully added {key} to {i}th index.")

    def search(self, key):
        i = self.HashFunction(key)
        original_index = i
        while self.table[i] is not None:
            if self.table[i] == key:
                print(f"[Search] {key} found at {i}th index.")
                return
            i = (i+1)%self.size
            if i == original_index:
                print(f"[Search] Hash table overflow!")
        print(f"[Search] {key} not found!")

    def delete(self, key):
        i = self.HashFunction(key)
        original_index = i
        while self.table[i] is not None:
            if self.table[i] == key:
                self.table[i] = None
                print(f"[Deletion] Successfully deleted {key} from {i}th index.")
                return
            i = (i+1)%self.size
            if i == original_index:
                print(f"[Deletion] Hash Table overflow!")
                return
        print(f"[Deletion] {key} not found to delete!")

    def display(self):
        print("\n Hash Table")
        for i in range(self.size):
            print(f"{i}: {self.table[i]}")
```

```python
obj = HashTableOpenAddressing()
obj.search(10)
obj.insert(10)
obj.search(10)
obj.insert(34)
obj.insert(44)
obj.delete(55)
obj.insert(55)
obj.search(55)
obj.delete(55)
obj.display()
```
