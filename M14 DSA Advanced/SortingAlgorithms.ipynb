# Sorting Algorithms

## Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        isSwapped = False
        for j in range(n-1-i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                isSwapped = True
        if not isSwapped: break
    return arr

print(bubble_sort([2]))
print(bubble_sort([2,3,5,1,7,4]))
print(bubble_sort(np.random.randint(low=10,high=100,size=10)))
```

## Insertion Sort

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i-1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

print(insertion_sort([]))
print(insertion_sort([3]))
print(insertion_sort([2,3,5,1,7,4]))
print(insertion_sort(np.random.randint(low=10,high=100,size=10)))
```
