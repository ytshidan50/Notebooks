# Binary Tree

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

root = Node(15)
root.left = Node(10)
root.right = Node(20)
root.left.left = Node(7)
root.left.right = Node(16)
root.right.left = Node(17)
root.right.right = Node(23)
```

## Preorder

```python
def preorder(node):
    if node:
        print(node.val)
        preorder(node.left)
        preorder(node.right)
preorder(root)
```

## Inorder

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.val)
        inorder(node.right)
inorder(root)
```

## Postorder

```python
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.val)
postorder(root)
```

# Binary Search Tree

### Creation

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
```

### Insertion

```python
def insert(root, val):
    if root is None:
        return Node(val)
    curr = root
    while curr != None:
        if val < curr.val:
            if curr.left != None:
                curr = curr.left
            else:
                curr.left = Node(val)
                break
        else:
            if curr.right != None:
                curr = curr.right
            else:
                curr.left = Node(val)
                break
    return root
```

```python
root = Node(7)
root = insert(root, 10)
inorder(root)
```

```python
root = insert(root, 5)
root = insert(root, 15)
preorder(root)
```
