```python
def paranthesis(s):
    st=[]
    if len(s) == 0: return True
    if s[0] == ")": return False
    for i in s:
        if i == "(":
            st.append(i)
        else:
            if st:
                st.pop()
    if not st:
        return True
    else:
        return False
print(paranthesis(""))
print(paranthesis(")()"))
print(paranthesis("(()"))
```

```python
class Node:
    def __init__(self, val):
        self.right = None
        self.left = None
        self.val = val


#        8
#      /   \
#     3     10
#    / \      \
#   1   6      14

root = Node(8)
root.left = Node(3)
root.right = Node(10)
root.left.left = Node(1)
root.left.right = Node(6)
root.right.right = Node(14)

def preorder(node):
    if node:
        print(node.val)
        preorder(node.left)
        preorder(node.right)

preorder(root)
```

```python
def insert(root, val):
    current = root
    while current:
        if current.val < val:
            if current.right is not None:
                current = current.right
            else:
                current.right = Node(val)
                break
        else:
            if current.left is not None:
                current = current.left
            else:
                current.left = Node(val)
                break
    return root

root = insert(root, 5)
preorder(root)
```
