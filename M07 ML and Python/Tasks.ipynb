```python
c=int(input("Enter the length of the array: "))
a=[input(f"Enter the value for pos {i+1}: ") for i in range(c)]
for i in range(c):
    for j in range(i+1,c):
        if a[j]>a[i]:
            temp=a[i]
            a[i]=a[j]
            a[j]=temp
print("The sorted array: ",end='')
[print(a[i], end=', ' if i!=c-1 else '') for i in range(c)]
print()
```

### User Defined Function:

```python
def isPrime(n):
    if n<=1: return 0
    for i in range(2,int(n/2)+1):
        if n%i==0:
            return 0
    return 1

if isPrime(int(input("Enter a number to check whether a number is prime or not: "))):
    print("The entered number is a prime.")
else:
    print("The entered number is not a prime!")
```

```python
from abc import ABC, abstractmethod

class Calculator:
    @abstractmethod
    def add(self):
        pass
    @abstractmethod
    def sub(self):
        pass
    @abstractmethod
    def mul(self):
        pass
    @abstractmethod
    def div(self):
        pass

class Calc(Calculator):
    def __init__(self,num1,num2):
        self.num1=num1
        self.num2=num2

    def add(self):
        return self.num1+self.num2

    def sub(self):
        return self.num1-self.num2

    def mul(self):
        return self.num1*self.num2

    def div(self):
        return self.num1/self.num2

c=Calc(10,3)
print(c.add())
print(c.sub())
print(c.mul())
print(c.div())
```

## 22. Write a program to create Class with name and account number and implement get and set, with property decorator and making account number and name private.

```python
class Bank:
    def __init__(self, name, account_number):
        self.__account_number=account_number
        self.__name=name

    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self,name):
        self.__name=name

    @property
    def account_number(self):
        return self.__account_number

    @account_number.setter
    def account_number(self,account_number):
        self.__account_number=account_number

b=Bank("Shidan",322)
print("Name:",b.name,"\nAccount Number:",b.account_number,end="\n\n")
b.name="Shidan V"
b.account_number=10322
print("Name:",b.name,"\nAccount Number:",b.account_number)
```

## 23. Write a function to calculate the sum of all numbers passed as its arguments. 
    a. Your function should be called sum_numbers and should define a single variable argument (i.e. a star argument) that will get the values to sum.
    b. Test the function with the following values:
| Values | Result |
| --- | --- |
| 1, 2, 3 | 6 |
| 8, 20, 2 | 30 |
| 12.5, 3.147, 98.1 | 113.747 |
| 1.1, 2.2, 5.5 | 8.8 |

```python
def sum_numbers(*args):
    s=0
    for x in args:
        s+=x
    return s

print(sum_numbers(1, 2, 3))
print(sum_numbers(8, 20, 2))
print(sum_numbers(12.5, 3.147, 98.1))
print(sum_numbers(1.1, 2.2, 5.5))
```
