```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def arrayToLinkedList(self, arr):
        if not arr:
            return None
        head = ListNode(arr[0])
        current = head
        for i in range(1, len(arr)):
            current.next = ListNode(arr[i])
            current = current.next
        return head

    def traverseForward(self, head):
        current = head
        while current:
            print(current.val)
            current = current.next

    def reverseLL(self, head):
        prev = None
        current = head
        while current:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        return prev

    def mergeDigits(self, head):
        current = head
        m=0
        while current:
            m = m * 10 + current.val
            current = current.next
        return m

    def addTwoNumbers(self):
        o = Solution()
        l1 = o.arrayToLinkedList([2, 4, 3])
        l2 = o.arrayToLinkedList([5, 6, 4])
        

        reversed_l1 = o.reverseLL(l1)
        reversed_l2 = o.reverseLL(l2)

        s = o.mergeDigits(reversed_l1) + o.mergeDigits(reversed_l2)
        










s=Solution()
s.addTwoNumbers()
```
