## Linked List Manual

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

node1 = Node(30)
node2 = Node(23)
node1.next = node2
node3 = Node(3)
node2.next = node3
node4 = Node(5)
node3.next = node4
head = node1
current = head
for i in range(1,5):
    print(current.data)
    current = current.next
```

## Linked List From Array

```python
def traverseForward(current):
    while current:
        print(current.data)
        current = current.next

def traverseBackward(current):
    while current:
        print(current.data)
        current = current.prev
```

```python
arr = [22,36,36,22,45,35,65]

class Node:
    def __init__(self, val):
        self.data = val
        self.next = None

head = Node(arr[0])
current = head
for i in range(1, len(arr)):
    new_node = Node(arr[i])
    current.next = new_node
    current = new_node

current = head
traverseForward(current)
```

## Reverse a Linked List.

```python
def reverseLL(head):
    current = head
    prev = None
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    return prev

current = reverseLL(head)
traverseForward(current)
```

## Create doubly linked list

```python
a = [4,6,8,5,3,5,8,3,7,4]

class Node:
    def __init__(self, val):
        self.data = val
        self.next = None
        self.prev = None


head = Node(a[0])
current = head
for i in range(1,len(a)):
    new_node = Node(a[i])
    new_node.prev = current
    current.next = new_node
    current = current.next
tail = current

current = head
traverseForward(current)
current = tail
traverseBackward(current)
```

## Add node begining

```python
def AddNodeBegining(head, node):
    node.next = head
    return node
```

```python
current = AddNodeBegining(head, Node(20398457))
traverseForward(current)
```

## Add node at end

```python
def AddNodeEnd(head, node):
    current = head
    while current.next:
        current = current.next
    current.next = node
    return head
```

```python
current = AddNodeEnd(head, Node(72232222222))
traverseForward(current)
```

## Add a node next to a value

```python
def AddNodeNextToValue(head, node, value):
    current = head
    while current:
        if current.data == value:
            node.next = current.next
            current.next = node
            return head
        current = current.next
    return head
```

```python
current = AddNodeNextToValue(head, Node(111), 8)
traverseForward(head)
```
