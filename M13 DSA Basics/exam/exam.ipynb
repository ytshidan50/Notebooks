```python
# fibinaccy

count = int(input("Enter the count: "))
ar=[1,1]
def fib(n1, n2, count):
    global ar
    if count ==0:
        return
    ar.append(n1+n2)

    fib(n2, n1+n2, count-1)

fib(1, 1, 5)
print(ar)
```

```python
l = [5]

for i in range(len(l)-1):
    for j in range(i+1, len(l)):
        if l[i]>l[j]:
            temp = l[i]
            l[i] = l[j]
            l[j] = temp
print(l)


```

```python
left = 0
right = len(l)-1
target = 5
while left < right:
    mid = left+right//2
    if l[mid] == target:
        print(mid)
        break
    elif l[left] < target:
        left += 1
    elif l[right] > target:
        right -= 1
    
```
