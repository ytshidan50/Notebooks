```python
def su(arr, s=0):
    if len(arr) == 0:
        return s
    return su(arr[1:], s+arr[0])

arr = [1,2,3,4,5]
print(su(arr))
```

```python
def revstr(st, s=""):
    if len(st) == 0:
        return s
    return revstr(st[1:], s+st[0])

print(revstr("Hello World"))
```

```python
class LinkNode:
    def __init__(self, val):
        self.val= val
        self.next = None
        self.prev = None

node1 = LinkNode(39)
node2 = LinkNode(84)
node3 = LinkNode(34)

node1.next = node2
node2.next = node3
node2.prev = node1
node3.prev = node2

head = node1

def traverseForward(head):
    current = head
    count = 0
    while current:
        print(current.val)
        current = current.next
        count +=1
    return count

def traverseBackward(tail):
    current = tail
    while tail:
        print(current.val)
        current = current.prev

def findValueByPos(head, pos):
    current = head
    c = 0
    while current:
        c+=1
        if c == pos:
            return current.val
        current = current.next
```

```python
cur = head
count = traverseForward(cur)
print(findValueByPos(cur, count//2+1))
```
